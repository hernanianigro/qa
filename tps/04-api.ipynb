##### Hacer una petición a https://httpbin.org/redirect/3 y analizar:
* historial (history)
* URL final (url)

```python
import requests

url = "https://httpbin.org/redirect/3"
response = requests.get(url)

print("HISTORIAL DE REDIRECCIONES:")
for i, r in enumerate(response.history, start=1):
    print(f"{i}. Estado: {r.status_code} - URL: {r.url}")

print("\nURL FINAL:")
print(response.url)

print("\nCÓDIGO DE ESTADO FINAL:")
print(response.status_code)
```

##### Hacer una petición a una URL inexistente y usar raise_for_status() para capturar el error.

```python
import requests

url = "https://httpbin.org/status/404"

try:
    response = requests.get(url)
    response.raise_for_status()
    print("Petición realizada correctamente")

except requests.exceptions.HTTPError as error:
    print("Se produjo un error HTTP:")
    print(error)
```

##### Descargar un archivo grande (https://httpbin.org/stream/5) y procesa su contenido línea por línea con iter_lines().

```python
import requests

url = "https://httpbin.org/stream/5"

response = requests.get(url, stream=True)

response.raise_for_status()

for linea in response.iter_lines():
    if linea:
        print(linea.decode("utf-8"))
```

##### Crear una función que consulte una API pública (por ejemplo, https://jsonplaceholder.typicode.com/posts) y:
* Obtenga todos los posts
* Muestre los títulos
* Permita enviar un nuevo post con POST

```python
import requests

BASE_URL = "https://jsonplaceholder.typicode.com/posts"

def obtener_posts():
    response = requests.get(BASE_URL)
    response.raise_for_status()

    posts = response.json()

    print("Títulos de los posts:\n")

    for post in posts:
        print(f"- {post['title']}")

    return posts


def crear_post(titulo, contenido, usuario_id):
    nuevo_post = {
        "title": titulo,
        "body": contenido,
        "userId": usuario_id
    }

    response = requests.post(BASE_URL, json=nuevo_post)
    response.raise_for_status()

    post_creado = response.json()

    print("\nPost creado correctamente:")
    print(f"ID: {post_creado['id']}")
    print(f"Título: {post_creado['title']}")
    print(f"Contenido: {post_creado['body']}")

    return post_creado


obtener_posts()
crear_post(
    "Mi nuevo post",
    "Este es el contenido de mi nuevo post.",
    1
)
```

##### Crear una función que utilizando la API swapi (https://swapi.dev/) muestre la lista de personajes, cada vez que muestre la información de una página preguntará si quiere seguir mostrando más personajes. A continuación pedirá el nombre de un personaje y te dará información sobre él.

```python
import requests

BASE_URL = "https://swapi.dev/api/people/"


def mostrar_personajes():
    url = BASE_URL

    while url:
        response = requests.get(url)
        response.raise_for_status()

        datos = response.json()

        print("\n--- PERSONAJES ---")

        for personaje in datos["results"]:
            print(f"- {personaje['name']}")

        url_siguiente = datos["next"]

        if url_siguiente:
            continuar = input(
                "\n¿Quieres mostrar más personajes? (s/n): "
            ).lower()

            if continuar != "s":
                break

            url = url_siguiente
        else:
            print("\nNo hay más personajes.")
            break


def buscar_personaje():
    nombre = input("\nIntroduce el nombre del personaje: ")

    params = {
        "search": nombre
    }

    response = requests.get(BASE_URL, params=params)
    response.raise_for_status()

    datos = response.json()

    if not datos["results"]:
        print("\nNo se encontró ningún personaje.")
        return

    for personaje in datos["results"]:
        print("\n--- INFORMACIÓN DEL PERSONAJE ---")
        print(f"Nombre: {personaje['name']}")
        print(f"Altura: {personaje['height']} cm")
        print(f"Peso: {personaje['mass']} kg")
        print(f"Color de pelo: {personaje['hair_color']}")
        print(f"Color de ojos: {personaje['eye_color']}")
        print(f"Género: {personaje['gender']}")
        print(f"Año de nacimiento: {personaje['birth_year']}")


def main():
    mostrar_personajes()

    buscar_personaje()


if __name__ == "__main__":
    main()
```

##### Trabajar con una API pública que le permita obtener datos, con esos datos responder alguna pregunta de interés para procesarlos (usar pandas) y luego hacer una visualización (usar matplotlib).

```python
import requests
import pandas as pd
import matplotlib.pyplot as plt


# Obtengo datos de la API de github

url = "https://api.github.com/search/repositories"

params = {
    "q": "stars:>10000",
    "sort": "stars",
    "order": "desc",
    "per_page": 100
}

response = requests.get(url, params=params)
response.raise_for_status()

datos = response.json()

print(f"Repositorios obtenidos: {len(datos['items'])}")


# Creo un dataframe con pandas

repositorios = []

for repo in datos["items"]:
    repositorios.append({
        "nombre": repo["full_name"],
        "estrellas": repo["stargazers_count"],
        "lenguaje": repo["language"]
    })

df = pd.DataFrame(repositorios)

print("\nPrimeros repositorios:")
print(df.head())


# Proceso los datos

df = df.dropna(subset=["lenguaje"])

lenguajes = df["lenguaje"].value_counts()

print("\nRepositorios por lenguaje:")
print(lenguajes)


```
