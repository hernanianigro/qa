Imprimir por pantalla los números del 1 al 100 pero considerando lo siguiente: a) Si el número es divisible por 3 se debe imprimir "Fizz". b) Si el número es divisible por 5 se debe imprimir "Buzz". c) Si el número es divisible por 3 y por 5 se debe imprimir "FizzBuzz"

```python
def esDivisiblePor3(numero):
    return numero % 3 == 0

def esDivisiblePor5(numero):
    return numero % 5 == 0

def fizzBuzz(numero):

    salida = None

    if esDivisiblePor3(numero) and esDivisiblePor5(numero):
        salida = "FizzBuzz"
    elif esDivisiblePor3(numero):
        salida = "Fizz"
    elif esDivisiblePor5(numero):
        salida = "Buzz"
    else:
        salida = numero

    return salida

for numero in range(1, 101):
    print(fizzBuzz(numero))
```

```python
import unittest
class fizzBuzz(unittest.TestCase):
  def test_fizzBuzz_3(self):
    self.assertEqual("Fizz",fizzBuzz(3))

  def test_fizzBuzz_5(self):
    self.assertEqual("Buzz", fizzBuzz(5))

  def test_fizzBuzz_15(self):
    self.assertEqual("FizzBuzz",fizzBuzz(15))
  
  def test_fizzBuzz_4(self):
    self.assertEqual(4, fizzBuzz(4))
```

Los números expansivos se definen de la siguiente forma:
* El primer número expansivo es el 1
* Dado un número expansivo x, para calcular el siguiente se ha de sustituir cada grupo de n cifras consecutivas iguales que aparezcan en x (por ejemplo, 3333) por número n seguido de la cifra que se repite (en el ejemplo 43)
* Así, los primeros números expansivos son:
* 1 -> "un uno": 11
* 11 -> "dos unos": 21
* 21 -> "un dos, un uno": 1211
* 1211 -> "un uno, un dos, dos uno": 111221

Se pide programar: 
1. una función "expand" que, dado un arreglo con las cifras de un número expansivo genere el siguiente
2. una función "list2num" que, dado un arreglo con las cifras de un número expansivo, devuelva el número correspondiente

```python
def expand(arreglo):
  salida = []
  numero = arreglo[0]
  cantidad = 1

  for i in range(1, len(arreglo)):
      if arreglo[i] == numero:
        cantidad += 1
      else:
        salida.append(cantidad)
        salida.append(numero)

        numero = arreglo[i]
        cantidad = 1
  salida.append(cantidad)
  salida.append(numero)
  return salida
```

```python
import unittest
class ExpandTest(unittest.TestCase):
  def test_expand_(self):
    self.assertEqual([2, 1, 1, 2, 2, 1], expand([1, 1, 2, 1, 1]))

  def test_expand1_(self):
    self.assertEqual([2,1], expand([1,1]))
    
  def test_expand2_(self):
    self.assertEqual([3, 1], expand([1, 1, 1]))
```

```python
resultado = expand([1, 1, 2, 1, 1])
print(resultado)
```

```python
def list2num(arreglo):
  numero = 0
  for cifra in arreglo:
    numero = numero * 10 + cifra
  return numero
```

```python
import unittest
class List2NumTest(unittest.TestCase):
  def test_list2num_11(self):
    self.assertEqual(11, list2num([1, 1]))

  def test_list2num_1211(self):
    self.assertEqual(1211, list2num([1, 2, 1,1]))
```

```python
resultado_numero = list2num([1, 2, 3, 4])
print(resultado_numero)
```
